```python
import os

from langchain_experimental.text_splitter import SemanticChunker
from langchain_openai import OpenAIEmbeddings

if not os.getenv("OPENAI_API_KEY"):
    raise ValueError("Set OPENAI_API_KEY before running this example.")

embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small"
)

splitter = SemanticChunker(
    embeddings,
    breakpoint_threshold_type="percentile",
    breakpoint_threshold_amount=90,
)

text = """
The employee submits a document for approval.
The quality reviewer checks the document and approves or rejects it.
The system stores the final document version.

The application uses PostgreSQL to store workflow records.
The database is backed up every night.
"""

chunks = splitter.create_documents([text])

for i, chunk in enumerate(chunks, start=1):
    print(f"\n--- Chunk {i} ---")
    print(chunk.page_content)
```
