How does RAG compare two vectors?
RAG systems commonly use a similarity or distance metric to rank document vectors against the query vector.

Three important metrics are:

Cosine similarity

Cosine similarity measures the cosine of the angle between two nonzero vectors. A score of 1 means they point in the same direction, 0 means they are perpendicular, and -1 means they point in opposite directions.



Dot product

Multiplies corresponding vector values and adds the results. The score depends on both direction and vector magnitudes unless vectors are normalized.




Euclidean distance (L2)

Measures the straight-line distance between two vectors. A smaller distance means the vectors are closer in that vector space.

5. Symmetric vs. asymmetric search
This is an important concept for RAG interviews.

Symmetric search is when the query and the documents have similar forms. For example, finding a sentence similar to another sentence.

Asymmetric search is when a short query needs to retrieve a longer passage or document.

Example:
- Query: “How can I recover my account?”
- Document: “To recover your account, open the sign-in page, select Forgot Password, and follow the verification steps.”
This is asymmetric search because the query is short and the relevant document is longer.

8. How to evaluate an embedding model in a real project
Don't evaluate the model only by checking whether its vectors look correct. Embedding vectors are not usually interpretable by inspection.

Instead, create a representative evaluation dataset:
1. Collect realistic user queries.
2. Label which document chunks are relevant to each query.
3. Generate embeddings using candidate models.
4. Retrieve the top \(k\) chunks for each query.
5. Compare retrieval metrics, latency, and cost.
Useful metrics include:
- Recall@k: How often relevant documents appear among the top \(k\) results.
- Precision@k: What fraction of the top \(k\) results are relevant.
- MRR (Mean Reciprocal Rank): How highly the first relevant result appears in the ranking.
