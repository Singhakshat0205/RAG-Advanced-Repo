```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document

# 1. Initialize the embedding model
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# 2. Prepare documents
documents = [
    Document(
        page_content="To reset your password, open the login page "
                     "and select Forgot Password.",
        metadata={"topic": "password"}
    ),
    Document(
        page_content="To recover your account, complete the identity "
                     "verification process.",
        metadata={"topic": "account recovery"}
    ),
    Document(
        page_content="Chocolate cake recipes commonly use flour, "
                     "sugar, and cocoa powder.",
        metadata={"topic": "cooking"}
    ),
]

# 3. Index documents: generate and store their vectors
vector_store = FAISS.from_documents(
    documents,
    embeddings
)

# 4. Receive a user query
query = "How can I regain access to my account?"

# 5. Retrieve similar document chunks
results = vector_store.similarity_search(
    query,
    k=2
)

for doc in results:
    print("Content:", doc.page_content)
    print("Metadata:", doc.metadata)
    print("---")
```
