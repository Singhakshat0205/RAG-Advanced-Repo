```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = """
Approval Workflow

If approval fails, verify the assigned reviewer.
If the reviewer is correct, check the current workflow state.
If the workflow is still blocked, inspect the error message.

Document Control

Verify that the document version is current.
Confirm that the required review has been completed.
"""

configs = [
    {"chunk_size": 100, "chunk_overlap": 0},
    {"chunk_size": 100, "chunk_overlap": 20},
    {"chunk_size": 200, "chunk_overlap": 30},
]

for config in configs:
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=config["chunk_size"],
        chunk_overlap=config["chunk_overlap"],
        separators=["\n\n", "\n", " ", ""],
    )

    chunks = splitter.split_text(text)

    print(f"\nConfiguration: {config}")
    print("Number of chunks:", len(chunks))

    for i, chunk in enumerate(chunks, start=1):
        print(f"\nChunk {i} ({len(chunk)} characters):")
        print(chunk)
```
