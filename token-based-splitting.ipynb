```python
from langchain_text_splitters import TokenTextSplitter
```

```python
text = """
When an approval workflow fails, check the assigned reviewer.
Then inspect the workflow state and the relevant error message.
Verify that the document version and required approvals are correct.
"""

splitter = TokenTextSplitter(
    encoding_name="cl100k_base",
    chunk_size=50,
    chunk_overlap=15,
)

chunks = splitter.split_text(text)

for i, chunk in enumerate(chunks, start=1):
    print(f"\n--- Chunk {i} ---")
    print(chunk)
print(chunks)
```
