```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    encode_kwargs={"normalize_embeddings": True},
)

documents = [
    "You can reset your password from the login page.",
    "Chocolate cake requires flour and cocoa powder.",
]

document_vectors = embeddings.embed_documents(documents)

query_vector = embeddings.embed_query(
    "How can I recover my password?"
)

print("Number of document vectors:", len(document_vectors))
print("Vector dimensions:", len(document_vectors[0]))
print("Query vector dimensions:", len(query_vector))
```
